# bmk Quickstart

Install the package (from PyPI or GitHub), then try the CLI and API examples below.

## 1) Install latest from GitHub (default)

```python
import sys

# Install uv into this kernel's environment first: a bare `!uv` needs uv on
# PATH, which a reader running this notebook from a plain Python will not have.
!{sys.executable} -m pip install --quiet --upgrade uv
assert _exit_code == 0, "installing uv failed"

# install bmk from GitHub (default)
!{sys.executable} -m uv pip install --python {sys.executable} --upgrade git+https://github.com/bitranox/bmk.git
assert _exit_code == 0, "installing bmk failed"
```

## 2) CLI: version and help

```python
!bmk --version
assert _exit_code == 0, "bmk --version failed"

!bmk --help
assert _exit_code == 0, "bmk --help failed"

!bmk info
assert _exit_code == 0, "bmk info failed"
```

## 3) CLI fail without traceback

```python
!bmk fail
assert _exit_code == 1, "bmk fail is expected to exit 1"
```

## 4) CLI fail with traceback

```python
!bmk --traceback fail
assert _exit_code == 1, "bmk --traceback fail is expected to exit 1"
```

## 5) script fail without traceback

```python
!{sys.executable} -m bmk fail
assert _exit_code == 1, "python -m bmk fail is expected to exit 1"
```

## 6) script fail with traceback

```python
!{sys.executable} -m bmk --traceback fail
assert _exit_code == 1, "python -m bmk --traceback fail is expected to exit 1"
```
